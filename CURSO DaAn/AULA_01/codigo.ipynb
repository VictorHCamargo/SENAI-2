```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("loja.csv")
sns.set_theme(style="darkgrid")

vendas_estado = df.groupby("State")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(12,6))
sns.barplot(x=vendas_estado.index, y=vendas_estado.values)
plt.xticks(rotation=90)
plt.title("Vendas Totais por Estado")
plt.xlabel("Estado")
plt.ylabel("Vendas (R$)")
plt.show()


segmentos = df["Segment"].value_counts()

plt.figure(figsize=(7,7))
plt.pie(segmentos, labels=segmentos.index, autopct="%1.1f%%")
plt.title("Participação dos Segmentos")
plt.show()

plt.figure(figsize=(8,5))
plt.hist(df["Sales"], bins=25)
plt.title("Distribuição de Valores de Venda")
plt.xlabel("Valor da venda (R$)")
plt.ylabel("Frequência")
plt.show()

plt.figure(figsize=(8,6))
sns.boxplot(data=df, x="Region", y="Profit")
plt.title("Distribuição do Lucro por Região")
plt.xlabel("Região")
plt.ylabel("Lucro (R$)")
plt.show()

plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Category")
plt.title("Quantidade de Registros por Categoria")
plt.xlabel("Categoria")
plt.ylabel("Quantidade")
plt.show()
```
